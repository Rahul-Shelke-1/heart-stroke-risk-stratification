# ⭐ **4. EDA Checklist (Use Before Starting & While Working)**


### 🔹 **A. Setup**

* [✔] Import data
* [✔] Check schema, shape, columns
* [✔] Validate data types

### 🔹 **B. Data Quality**

### **B.1. Standardize Column Names**

These must be consistent across all datasets.

* [-] Trim spaces
* [✔] Convert to lowercase
* [-] Replace spaces/symbols with `_`
* [-] Remove special characters
* [-] Collapse multiple underscores

Example:
- `"Employee ID "` → `"employee_id"`
- `"Salary (USD)"` → `"salary_usd"`

---

### **B.2. Remove Unwanted / Corrupted Columns**

* [o] Drop ID-like columns not needed
* [-] Drop completely empty columns
* [-] Drop leakage columns (future info)

---

### **B.3. Fix Categorical Values (String Standardization)**

This is one of the biggest inconsistency areas.

#### ✔ Convert values to consistent case

* [ ] `Male` → `male`
* [ ] `YES` → `yes`

#### ✔ Fix spelling issues & variations

* [ ] `male, Male, M, m` → `male`
* [ ] Use mapping dict

#### ✔ Remove trailing/leading spaces

* [ ] `" Yes "` → `"yes"`

#### ✔ Strip symbols

`"₹50,000"` → `"50000"`

#### ✔ Map invalid or unknown values

* [ ] `"?"`, `"N.A"`, `"Missing"` → `None`

---

### **B.4. Handle Missing Values – But Keep Them Meaningful**

Before imputing anything, you MUST tag missing values consistently.

#### ✔ Standardize missing values

Convert everything to **None / np.nan**

#### ✔ IMPORTANT: Preserve nulls for analysis

This helps determine:

* MCAR
* MAR
* MNAR

#### ✔ Later you create an imputation strategy, but only *after* EDA.

---

### **B.5. Validate Numerical Columns**

#### A) Clean formatting

* [ ] Remove commas, currency symbols
* [ ] Convert to numeric types

#### B) Identify impossible values

* [ ] Salary: remove zeros / negative values
* [ ] Age: remove < 0 and > 120
* [ ] Ratings: clip outside valid ranges
* [ ] Convert strings like `"ten"` → numeric

#### C) Check unique values

* [ ] Identify single-value or constant columns → drop
* [ ] Detect numeric columns that are actually categorical (e.g., 0/1, 1/2/3)

---

### **B.6. Validate Date/Time Columns**

#### Tasks:

* [ ] Convert to proper datetime dtype
* [ ] Parse multiple date formats
* [ ] Handle invalid dates (`32/13/2025`)
* [ ] Extract useful features:

  * year
  * month
  * weekday
  * hour
  * days_since_start

#### Identify time-based inconsistencies:

* [ ] End date < start date
* [ ] Age calculation mismatches
* [ ] Missing timestamps

---

### **B.7. Validate Text Columns**

Text fields often contain dirty data.

#### Tasks:

* [ ] Lowercase
* [ ] Remove special characters
* [ ] Normalize unicode
* [ ] Remove HTML tags
* [ ] Standardize whitespace
* [ ] Remove/control emojis
* [ ] Spell correction (optional)
* [ ] Extract keyphrases (optional)

---

### **B.8. Fix Data Types**

After cleaning values:

* [ ] Convert categorical → `category` dtype
* [ ] Convert numeric strings → `float` or `int`
* [ ] Convert binary columns to 0/1 or yes/no
* [ ] Convert datetime strings → datetime
* [ ] Convert ID columns → string (not numeric)

---

### **B.9. Outlier Detection + Handling**

#### Methods:

* [ ] IQR
* [ ] Z-score
* [ ] Isolation Forest (optional)

Fix:

* [ ] Cap/clamp
* [ ] Log transform
* [ ] Remove if impossible

---

### **B.10. Integrity Checks After Cleaning**

* [ ] Shape: no accidental row loss
* [ ] Dtypes correct
* [ ] No corrupted values
* [ ] No duplicated rows
* [ ] No abnormal distributions
* [ ] All cleaning rules met
